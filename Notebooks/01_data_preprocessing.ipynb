## Data Dictionary

- **age**: Age in years  
- **sex**: Sex  
  - `1` = Male  
  - `0` = Female  

- **cp**: Chest pain type  
  - `1` = Typical angina  
  - `2` = Atypical angina  
  - `3` = Non-anginal pain  
  - `4` = Asymptomatic  

- **trestbps**: Resting blood pressure (mm Hg on admission to the hospital)  

- **chol**: Serum cholesterol (mg/dl)  

- **fbs**: Fasting blood sugar > 120 mg/dl  
  - `1` = True  
  - `0` = False  

- **restecg**: Resting electrocardiographic results  
  - `0` = Normal  
  - `1` = ST-T wave abnormality (T wave inversions and/or ST elevation or depression > 0.05 mV)  
  - `2` = Probable or definite left ventricular hypertrophy by Estes' criteria  

- **thalach**: Maximum heart rate achieved  

- **exang**: Exercise induced angina  
  - `1` = Yes  
  - `0` = No  

- **oldpeak**: ST depression induced by exercise relative to rest  

- **slope**: Slope of the peak exercise ST segment  
  - `1` = Upsloping  
  - `2` = Flat  
  - `3` = Downsloping  

- **ca**: Number of major vessels (0–3) colored by fluoroscopy  

- **thal**: Thalassemia status  
  - `3` = Normal  
  - `6` = Fixed defect  
  - `7` = Reversible defect  

- **num**: Diagnosis of heart disease (angiographic disease status)  
  - `0` = < 50% diameter narrowing  
  - `1` = > 50% diameter narrowing

## Import Dependencies

```python
from ucimlrepo import fetch_ucirepo 
import numpy as np
import pandas as pd
from sklearn.impute import SimpleImputer
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import StandardScaler, OneHotEncoder
import seaborn as sns
import matplotlib.pyplot as plt
import os
```

## Data understanding

```python
heart_disease = fetch_ucirepo(id=45) 

X = heart_disease.data.features 
y = heart_disease.data.targets
print(X.shape, y.shape)
```

```python
print(heart_disease.variables) 
```

```python
df = pd.concat([X, y], axis= 1)
df.head()
```

```python
df.info()
```

```python
df.describe()
```

## Handling Null values

```python
df.isnull().sum()
```

```python
imputer = SimpleImputer(strategy='median')      # Can't use mean as a strategy because it is categorical it will generate a new value
df_imputed = pd.DataFrame(
    imputer.fit_transform(df),
    columns=df.columns
)
df_imputed.isnull().sum()
```

## Data Encoding and Standardization

```python
target = "num"

numerical_cols = ["age", "trestbps", "chol", "thalach", "oldpeak", "ca"]

categorical_cols = ["sex", "cp", "fbs", "restecg", "exang", "slope", "thal"]
```

```python
# Building preprocessing pipeline
preprocessor = ColumnTransformer(
    transformers=[
        ('num', StandardScaler(), numerical_cols),
        ('cat', OneHotEncoder(), categorical_cols)
    ],
    remainder="drop")

df_clean = preprocessor.fit_transform(df_imputed)
```

```python
feature_names = numerical_cols + list(preprocessor.named_transformers_['cat'].get_feature_names_out(categorical_cols))

df_processed = pd.DataFrame(df_clean, columns=feature_names)
df_processed[target] = df_imputed[target].values

df_processed
```

```python
df_processed.columns
```

## EDA and Visualization

```python
# Histogram for Numerical columns
plt.figure(figsize=(14,10))
for i, col in enumerate(numerical_cols):
    plt.subplot(3, 2, i+1)
    sns.histplot(df_processed[col], bins=30, kde=True)
    plt.title(f'Histogram for {col}')
plt.tight_layout()
plt.show()
```

```python
# Boxplot for Numerical columns
plt.figure(figsize=(12,10))
for i, col in enumerate(numerical_cols):
    plt.subplot(3, 2, i+1)
    sns.boxplot(x=df_processed[col])
    plt.title(f"Boxplot for {col}")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(14, 10))
for i, col in enumerate(categorical_cols):
    plt.subplot(3, 3, i+1)
    unique = df_imputed[col].value_counts()
    count = unique.values
    categories = unique.index
    plt.pie(count, labels=categories, autopct="%1.1d%%", startangle=90)
    plt.xlabel(f"{col}")
    plt.ylabel('Density')
    plt.title(f"{col} Distribution Graph")
plt.subplots_adjust(hspace=0.8, wspace=0.3)
plt.show()
```

```python
# Countplot for categorical columns
plt.figure(figsize=(14, 10))
for i, col in enumerate(categorical_cols):
    plt.subplot(3, 3, i+1)
    sns.countplot(x=col, data=df_imputed)
    plt.xlabel(f"{col}")
    plt.ylabel('Density')
    plt.title(f"{col} Distribution Graph")
plt.subplots_adjust(hspace=0.8, wspace=0.3)
plt.show()
```

```python
plt.figure()
sns.countplot(x= target, data= df_imputed)
plt.xlabel("Target")
plt.ylabel("Density")
plt.title("Target distribution graph")
plt.show()
```

After observation of the target column there is inbalance in classes.

As said in data description: `It is integer valued from 0 (no presence) to 4. Experiments with the Cleveland database have concentrated on simply attempting to distinguish presence (values 1,2,3,4) from absence (value 0).`

So we can solve this problem by making the target value 0(no presence) and 1 (presence).

```python
df_processed["num"] = (df_processed["num"] > 0).astype(int)
df_processed["num"]. value_counts()
```

```python
plt.figure()
sns.countplot(x= df_processed["num"], data= df_imputed)
plt.xlabel("Target")
plt.ylabel("Density")
plt.title("Target distribution graph")
plt.show()
```

```python
# Correlation heatmap
plt.figure(figsize=(14,10))
sns.heatmap(df_processed.corr(), cmap="coolwarm", annot= True, fmt="0.2g")

plt.title("Correlation Matrix")
plt.tight_layout()
plt.show()
```
