## Import Dependencies

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.decomposition import PCA
```

## Load Dataset

```python
df = pd.read_csv("../Data/heart_disease_processed.csv")
df
```

```python
features = df.drop(columns=["num"])
target = df["num"]
```

## Applying PCA

```python
pca = PCA()
X_pca = pca.fit_transform(features)

explained_variance = pca.explained_variance_ratio_
explained_variance
```

## Visualize PCA to determine the optimal number of principal components 

```python
plt.figure(figsize=(8, 6))
plt.scatter(X_pca[:, 0], X_pca[:, 1], c= target, cmap='viridis')
plt.xlabel("First PC")
plt.ylabel("Second PC")
plt.title("PCA for dataset")
plt.colorbar(label='Target')
plt.tight_layout()
plt.show()
```

```python
explained_variance = pca.explained_variance_ratio_
cumulative_variance = np.cumsum(explained_variance)

plt.figure(figsize=(8,5))
plt.plot(range(1, len(explained_variance)+1), cumulative_variance, marker='o', color='red', label='Cumulative Variance')
plt.axhline(y=0.95, linestyle='--', label='95% Variance Threshold')
plt.xlabel('Principal Component')
plt.ylabel('Variance Ratio')
plt.title('Cumulative Variance by PCA Components')
plt.legend()
plt.tight_layout()
plt.show()
```

## Selecting best components optained by PCA

```python
n_components = np.argmax(cumulative_variance >= 0.95) + 1
print(f"Number of components above 95% threshold is{n_components}")
```

```python
pca = PCA(n_components=n_components)
X_pca = pca.fit_transform(features)
print(f"Shape of the Dataset after PCA is: {X_pca.shape}")
```

```python
pca_cols = [f'PC{i+1}' for i in range(n_components)]
df_pca = pd.DataFrame(X_pca, columns=pca_cols)
df_pca['Target'] = target
```

## Saving PCA data as CSV file

```python
save_path = "../Data/heart_disease_PCA.csv"
df_pca.to_csv(save_path, index=False)
print("File saved successfuly")
```
